```python
# q22 

arr5 = np.random.random((5, 5))
min = arr5.min()
max = arr5.max()

arr5 = (arr5 - min) / (max - min)

print(arr5)
```

```python
# q58

X = np.random.rand(5, 10)
Y = X - X.mean(axis=1, keepdims=True)

print(Y)
```

```python
# q25

arr7 = np.arange(11)
arr7[(3 < arr7) & (arr7 < 8)] *= -1

print(arr7)
```
