## 9. (Ex #69) How to get the diagonal of a dot product?

```python
A = np.random.uniform(0, 1, (5, 5))
B = np.random.uniform(0, 1, (5, 5))

# Slow way (computes full matrix, wastes memory):
slow = np.diag(A @ B)

# Fast way (avoids computing off-diagonal entries):
fast = np.einsum("ij,ji->i", A, B)

print(np.allclose(slow, fast))
```

## 10. (Ex #44) Convert cartesian coordinates (10x2 matrix) to polar coordinates

```python
Z = np.random.random((10, 2))
X, Y = Z[:, 0], Z[:, 1]
R = np.sqrt(X**2 + Y**2)
T = np.arctan2(Y, X)
print(R)
print(T)
```

## 11. (Ex #52) Given a (100,2) vector of coordinates, find point-to-point distances

```python
Z = np.random.random((100, 2))

# Method 1: broadcasting
X, Y = np.atleast_2d(Z[:, 0], Z[:, 1])
D = np.sqrt((X - X.T)**2 + (Y - Y.T)**2)

# Method 2 (faster, using scipy):
# import scipy.spatial
# D = scipy.spatial.distance.cdist(Z, Z)

print(D.shape)
print(D)
```

## 13. (Ex #42) Check if two random arrays are equal

```python
A = np.random.randint(0, 2, 5)
B = np.random.randint(0, 2, 5)

# Assuming identical shape, check element-wise equality with tolerance
equal = np.allclose(A, B)

# Strict check (same shape and same values, no tolerance):
equal_strict = np.array_equal(A, B)

print(equal, equal_strict)
```

## 15. (Ex #89) How to get the n largest values of an array

```python
Z = np.arange(10000)
np.random.shuffle(Z)
n = 5

# Fast (does not fully sort the array):
print(Z[np.argpartition(-Z, n)[:n]])
```

## 16. (Ex #65) Accumulate elements of vector X into array F, based on index list I

```python
X = [1, 2, 3, 4, 5, 6]
I = [1, 3, 9, 3, 4, 1]
F = np.zeros(10)
np.add.at(F, I, X)   # accumulates values at repeated indices correctly
print(F)
```

## 17. (Ex #67) Given a 4D array, get the sum over the last two axes at once

```python
A = np.random.randint(0, 10, (3, 4, 3, 4))

# Reshape to flatten the last two axes into one, then sum
sum_ = A.reshape(A.shape[:-2] + (-1,)).sum(axis=-1)
print(sum_.shape)
print(sum_)
```

## 19. (Ex #75) Compute averages using a sliding window over an array

```python
def moving_average(a, n=3):
    ret = np.cumsum(a, dtype=float)
    ret[n:] = ret[n:] - ret[:-n]
    return ret[n - 1:] / n

Z = np.arange(20)
print(moving_average(Z, n=3))
```

## 20. (Ex #100) Compute bootstrapped 95% confidence intervals for the mean of a 1D array

```python
X = np.random.randn(100)  # random 1D data
N = 1000                  # number of bootstrap samples

idx = np.random.randint(0, X.size, (N, X.size))
means = X[idx].mean(axis=1)
confint = np.percentile(means, [2.5, 97.5])
print(confint)
```
