```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

l1 = range(15)
l2 = [x*x*x for x in l1]
print(l1)
print(l2)
plt.plot(l1, l2)
plt.show()
```

```python
x = np.arange(2, 22, 2)
y = np.arange(2, 22, 2)
print(x)
print(y)
plt.title("Line Plot")
plt.xlabel("Range 2 - 20")
plt.ylabel("Range 2 - 20")
plt.plot(x, y)
plt.savefig("plot1.jpg")
plt.show()
```

```python
x = np.arange(0, 15, 0.2)
y1 = np.tan(x)
y2 = np.cos(x)
print(x)
print(y1)
print(y2)
plt.title("Tan and Cos Curve")
plt.xlabel("0 to 15 (0.2)")
plt.ylabel("-1 to 1")
plt.plot(x, y1)
plt.plot(x, y2)
plt.legend(['tan', 'cos'])
plt.savefig("plot2.jpg")
plt.show()
```

```python
a = np.arange(0, 50, 2)
b = np.arange(25, -25, -2)
c = np.arange(50, 0, -2)
print(a)
print(b)
print(c)
plt.figure(figsize=(6, 6))
plt.title("With Grid lines and sized")
plt.xlabel("Range 0 - 50")
plt.ylabel("Range 25 to -25")
plt.grid(True)
plt.xticks(np.linspace(0, 50, 11, endpoint=True))
plt.scatter(a, b, label="One", color="blue")
plt.scatter(a, c, label="Two", color="orange")
plt.legend(['A', 'B'])
plt.show()
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(1, 1, 1)
ax2 = ax1.twinx()
x = np.linspace(0., 5., 50)
y = [i**2 for i in x]
z = [i**3 for i in x]
ax1.plot(x, y, 'g-')
ax2.plot(x, z, 'm-')
ax1.set_ylabel('X^2', color='green')
ax2.set_ylabel('X^3', color='magenta')
ax1.set_xlabel('X')
plt.show()
```

```python
x = np.arange(0, 50, 2)
y = np.arange(25, -25, -2)
z = np.arange(50, 0, -2)
plt.plot(x, y, linewidth=4, c='#00FF00', linestyle=':')
plt.plot(x, z, c='r', marker='*', markersize=6)

a = np.linspace(-3.14, 3.14, 50)
print(a)
b = np.sin(a)
plt.plot(a, b, linewidth=4, c='#00FF00', linestyle=':')
plt.plot(a, b, c='r', marker='*', markersize=8)
plt.show()
```

```python
# seaborn
# - statistical analysis library for python
# - based on matplotlib.pyplot, uses plt for graphing
```

```python
import seaborn as sb
import matplotlib.pyplot as plt
print(sb.get_dataset_names())
```

```python
d = sb.load_dataset("planets")
m = d['method'].value_counts().nlargest(3).index
d = d[d['method'].isin(m)].copy()
d['dec'] = (d['year'] // 10) * 10
print(d.info())
print(d.head())
```

```python
plt.figure(figsize=(4, 4))
plt.scatter(d['year'], d['distance'])
plt.show()
```

```python
a = d[['number', 'orbital_period', 'mass', 'distance', 'year']].corr()
print(a)
sb.heatmap(a, annot=True)
plt.show()
```

```python
d2 = d.groupby(['year', 'method'], as_index=False).size()
print(d2.head())
sb.relplot(x="year", y="size", data=d2, kind="line", hue="method")
plt.show()
```

```python
sb.relplot(x="year", y="size", data=d2, kind="line", col="method")
plt.show()
```

```python
sb.relplot(data=d, x="orbital_period", y="mass", hue="method")
plt.show()
```

```python
d3 = d[d['number'] <= 2]
sb.relplot(data=d3, x="orbital_period", y="mass", hue="method", col="number")
plt.show()
```
