```python
# Synthetic weather data generation using normal distributions across major Indian cities.
import numpy as np
import pandas as pd

indian_cities = ['Surat', 'Nagpur', 'Indore', 'Bhopal', 'Patna', 'Vadodara', 'Ludhiana', 'Agra', 'Nashik', 'Ranchi', 'Guwahati', 'Coimbatore', 'Mysuru']
num_cities = len(indian_cities)

weather_data = {
    'Avg_Temperature_C': np.random.normal(28, 3.5, num_cities),
    'Humidity_Pct': np.random.normal(70, 6.0, num_cities),
    'Wind_Speed_KMH': np.random.normal(15, 2.5, num_cities)
}

city_weather_df = pd.DataFrame(data=weather_data, index=indian_cities)
print(city_weather_df)
```

```python
# Plotting 2D multi-cluster dataset with explicit labels and custom styling.
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import pandas as pd

X_data, y_data = make_blobs(n_samples=180, n_features=2, centers=4, random_state=11)
df_clusters = pd.DataFrame({'dim_1': X_data[:, 0], 'dim_2': X_data[:, 1], 'group': y_data})
palette = {0: 'teal', 1: 'coral', 2: 'slateblue', 3: 'goldenrod'}

fig, axis = plt.subplots()
for cluster_id, sub_df in df_clusters.groupby('group'):
    sub_df.plot(ax=axis, kind='scatter', x='dim_1', y='dim_2', label=f'Cluster {cluster_id}', color=palette[cluster_id])
plt.show()
```

```python
# Simulating non-linear semi-circular distribution pattern.
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
import pandas as pd

X_moon, y_moon = make_moons(n_samples=250, noise=0.08, random_state=3)
df_moon = pd.DataFrame({'x1': X_moon[:, 0], 'x2': X_moon[:, 1], 'target': y_moon})
moon_colors = {0: 'darkslategray', 1: 'orangered'}

fig, axis = plt.subplots()
for label, group_df in df_moon.groupby('target'):
    group_df.plot(ax=axis, kind='scatter', x='x1', y='x2', label=f'Class {label}', color=moon_colors[label])
plt.show()
```

```python
# Constructing nested circular dataset for binary classification evaluation.
from sklearn.datasets import make_circles
import matplotlib.pyplot as plt
import pandas as pd

X_ring, y_ring = make_circles(n_samples=200, noise=0.05, factor=0.4, random_state=17)
df_ring = pd.DataFrame({'coord_x': X_ring[:, 0], 'coord_y': X_ring[:, 1], 'class': y_ring})
ring_colors = {0: 'indigo', 1: 'gold'}

fig, axis = plt.subplots()
for class_id, group in df_ring.groupby('class'):
    group.plot(ax=axis, kind='scatter', x='coord_x', y='coord_y', label=f'Group {class_id}', color=ring_colors[class_id])
plt.show()
```

```python
# Exercise 1 - Custom 2-class Indian Wildlife distribution plot (Cranes vs. Bisons).
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

features_wild, labels_wild = make_blobs(n_samples=160, cluster_std=0.5, centers=[[3, 6], [7, 2]], random_state=25)
fig, ax = plt.subplots()
species_colors = ["teal", "darkorange"]
species_names = ["Cranes", "Bisons"]

for category in range(2):
    ax.scatter(features_wild[labels_wild == category, 0], features_wild[labels_wild == category, 1], c=species_colors[category], s=45, label=species_names[category])

ax.set(xlabel='Feature X', ylabel='Feature Y', title='Wildlife Distribution')
ax.legend(loc='upper right')
plt.show()
```

```python
# Binary cluster scatter plot for Transaction Fraud Detection (Valid vs. Fraud).
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

txn_data, txn_labels = make_blobs(n_samples=160, cluster_std=0.35, centers=[[2, 4], [5, 8]], random_state=19)
fig, ax = plt.subplots()
status_colors = ["seagreen", "crimson"]
status_labels = ["Valid Txn", "Fraud Txn"]

for status in range(2):
    ax.scatter(txn_data[txn_labels == status, 0], txn_data[txn_labels == status, 1], c=status_colors[status], s=45, label=status_labels[status])

ax.set(xlabel='Txn Amount Normalized', ylabel='Frequency Index', title='Transaction Classifications')
ax.legend(loc='upper left')
plt.show()
```

```python
# Multi-class combined shape dataset representing Indian Fauna distribution.
import numpy as np
import matplotlib.pyplot as plt
import sklearn.datasets as ds

circle_coords, circle_lbls = ds.make_circles(n_samples=120, shuffle=True, noise=0.05, random_state=30)
blob_centers = [[3, 4], [7, 3], [5, 8]]
blob_coords, blob_lbls = ds.make_blobs(n_samples=120, cluster_std=0.35, centers=blob_centers, random_state=14)

for idx in range(len(blob_centers) - 1, -1, -1):
    blob_lbls[blob_lbls == idx] = idx + 2

combined_labels = np.concatenate([circle_lbls, blob_lbls])
scaled_circle_coords = circle_coords * [1.3, 1.8] + [5, 5]
combined_coords = np.concatenate([scaled_circle_coords, blob_coords], axis=0)

fig, ax = plt.subplots()
fauna_colors = ["sienna", "midnightblue", "seagreen", "crimson", "purple"]
fauna_names = ["Sloth Bear", "Indian Wolf", "Barasingha", "Nilgai", "Indian Rhino"]

for fauna_id in range(len(blob_centers) + 2):
    ax.scatter(combined_coords[combined_labels == fauna_id, 0], combined_coords[combined_labels == fauna_id, 1], c=fauna_colors[fauna_id], s=40, label=fauna_names[fauna_id])

ax.legend(loc='upper right')
plt.show()
```
